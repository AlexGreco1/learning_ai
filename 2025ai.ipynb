What is AI
----------

Artificial Intelligence (AI) is the field of computer science focused on creating systems that perform tasks which normally require human intelligence. 
AI is no Chatgpt. AI is the field where there are many other subjects as you can observe in the next graphs.
















As you can observe in the diagrams these groups change over the time expanding and new groups appear.  

Key points:

- Definition: Methods and systems that perceive, reason, learn, and act to solve problems or accomplish goals.
- Main approaches: symbolic/logic-based methods, machine learning (statistical models), and deep learning (neural networks).
- Types: Narrow AI (task-specific), General AI (human-level, theoretical), Superintelligence (beyond human, speculative).
- Common applications: vision, speech, language, recommendation, planning, robotics, and data analysis.
- Limitations & concerns: dataset bias, robustness, interpretability, privacy, safety, and ethical impacts.

For a quick example: supervised ML learns a mapping from inputs to outputs using labeled data; deep learning uses layered neural networks to learn complex patterns from large datasets.

## AI ethic and considerations

AI has many applications, however is indeterministic and could be less predictable and reliable but more humanized. When we use the AI in active or passive way, is different from automation task. AI can be perceived as intelligence, but their utilization it might have some consideration.

1. **Security and Privacy**  
- **Perspective A — Data Minimization:** Limit the data you provide to AI tools. Avoid sharing sensitive identifiers.  
  **Source:** [NIST AI Risk Management Framework](https://www.nist.gov/itl/ai-risk-management-framework)  
- **Perspective B — Secure Model Deployment:** Use sandboxing, local execution, access controls, and monitoring when running AI locally (e.g., LLMs in containers).  
  **Source:** [OWASP AI Security & Privacy Guide](https://owasp.org/www-project-ai-security-and-privacy-guide/)  
- **Perspective C — Supply Chain Trust:** Check the provenance of models, datasets, and containers to avoid tampered artifacts.  
  **Source:** [ISO/IEC 5338:2024 AI Engineering](https://www.iso.org/standard/81118.html)  

2. **Human Oversight and Critical Thinking**  
- **Perspective A — Human-in-the-Loop:** AI outputs should be reviewed, validated, and cross-verified, especially for decisions involving finance, legal, medical, or safety.  
  **Source:** [EU AI Act Overview](https://digital-strategy.ec.europa.eu/en/policies/european-approach-artificial-intelligence)  
- **Perspective B — Error Awareness:** LLMs can hallucinate facts, misinterpret context, or fabricate citations; treat outputs as drafts, not truth.  
  **Source:** [Stanford HAI Hallucination Study](https://hai.stanford.edu/research)  
- **Perspective C — Cognitive Support, Not Replacement:** Use AI for brainstorming, structuring, and accelerating tasks rather than replacing human evaluation.  
  **Source:** [MIT Human-AI Collaboration Lab](https://iccc.mit.edu)  

3. **Transparency and Explainability**  
- **Perspective A — Traceability:** Prefer models that allow logs, prompts, datasets, and version control for auditing.  
  **Source:** [NIST Explainable AI Program](https://www.nist.gov/programs-projects/explainable-artificial-intelligence)  
- **Perspective B — Layered Explanations:** Provide high-level reasoning to end users and deeper technical details for expert review.  
  **Source:** [DARPA XAI](https://www.darpa.mil/program/explainable-artificial-intelligence)  
- **Perspective C — Documented Limitations:** Record known failure modes and boundary conditions of your AI system.  

4. **Fairness and Bias**  
- **Perspective A — Dataset Bias Checks:** Review data sources for demographic imbalance or unintended patterns.  
  **Source:** [IBM AI Fairness 360](https://aif360.mybluemix.net)  
- **Perspective B — Model Bias Monitoring:** Continuously test outputs for skewed recommendations or selective errors.  
  **Source:** [Google Responsible AI Practices](https://ai.google/responsibility/)  
- **Perspective C — Sociotechnical Context:** Bias is not only technical; consider real-world institutions, policies, and outcomes affected by the model.  

5. **Reliability and Robustness**  
- **Perspective A — Adversarial Resistance:** Protect against prompt injection, jailbreaks, and adversarial inputs.  
  **Source:** [OpenAI Prompt Injection Guidance](https://platform.openai.com/docs/guides/prompt-injection)  
- **Perspective B — Stress Testing:** Test models under extreme, ambiguous, or multilingual inputs.  
  **Source:** [ML Testing Best Practices](https://arxiv.org/abs/2107.03051)  
- **Perspective C — Fail-Safe Design:** Ensure system defaults to safe behavior when uncertain.  

6. **Accountability and Governance**  
- **Perspective A — Clear Responsibility:** Define who owns the model, data, and outputs and who is responsible for errors.  
  **Source:** [OECD AI Principles](https://oecd.ai/en/ai-principles)  
- **Perspective B — Policy Integration:** Adopt internal rules for acceptable use, logging, retention, and review cycles.  
- **Perspective C — Documentation:** Use model cards, data sheets, and risk assessments.


## Prompting engineering

In LLM world, prompting is the way we ask question or task to the models. This is important because depending of the question we could have some different answer, so if we want to have a little more control over the output response of the models we need to be more precise in the pormpting. Some of the key considearions are:

1. **Clarity and Specificity**  
- **Perspective A — Clear Intent:** State the goal, constraints, audience, and format to reduce ambiguity.  
  **Source:** [OpenAI Prompting Guide](https://platform.openai.com/docs/guides/prompting)  
- **Perspective B — Use Structures:** Bullet points, numbered steps, and examples increase accuracy and reduce hallucinations.  
  **Source:** [Microsoft Prompt Engineering Guidelines](https://learn.microsoft.com/ai/concepts/prompt-engineering)  
- **Perspective C — Explicit Boundaries:** Define what to exclude (e.g., “no code,” “no opinions”), which improves precision.  

2. **Progressive Refinement**  
- **Perspective A — Iterative Prompting:** Start with a rough prompt, review, then refine based on gaps.  
  **Source:** [Stanford CRFM Prompting Research](https://crfm.stanford.edu)  
- **Perspective B — Few-Shot Examples:** Provide samples of the desired output format or reasoning style.  
  **Source:** [Brown et al., GPT-3 Paper](https://arxiv.org/abs/2005.14165)  
- **Perspective C — Decomposition:** Break complex tasks into smaller steps and combine outputs.  

3. **Role and Context Setting**  
- **Perspective A — Role Priming:** Define an expert role (“act as a cybersecurity analyst”) to guide domain depth.  
  **Source:** [Anthropic Prompt Craft Best Practices](https://docs.anthropic.com)  
- **Perspective B — Persistent Context:** Provide background, prior attempts, constraints, and environment so the model remains aligned.  
- **Perspective C — Multi-Tool Context:** If using LLMs with retrieval or agents, include instructions on when to search, summarize, or verify.  

4. **Precision Through Constraints**  
- **Perspective A — Output Constraints:** Specify length, style, tone, structure, or technical level.  
  **Source:** [Google Prompt Design Guide](https://developers.generativeai.google/guide/prompting)  
- **Perspective B — Format Constraints:** Request JSON, code blocks, schemas, or tables when needed for downstream automation.  
- **Perspective C — Safety Constraints:** Declare prohibited content or require source links for factual tasks.  

5. **Critical Thinking and Error Reduction**  
- **Perspective A — Verification Prompts:** Add instructions like “list uncertainties,” “challenge assumptions,” or “validate against X.”  
  **Source:** [Berkeley CoT & Self-Consistency Research](https://arxiv.org/abs/2203.11171)  
- **Perspective B — Alternative Paths:** Ask the model to provide 3–5 different solution approaches.  
- **Perspective C — Reflection and Re-evaluation:** Request correction passes or cross-checking.  

6. **Advanced Techniques**  
- **Perspective A — Chain-of-Thought (High-Level):** Request structured reasoning summaries without exposing raw internal reasoning.  
  **Source:** [Wei et al., CoT Reasoning](https://arxiv.org/abs/2201.11903)  
- **Perspective B — ReAct (Reason + Act) Pattern for Agents:** Guide the model to alternate reasoning and tool use.  
  **Source:** [ReAct Paper](https://arxiv.org/abs/2210.03629)  
- **Perspective C — Retrieval-Augmented Prompts:** Include relevant documents or embeddings for accuracy.  
  **Source:** [RAG Survey](https://arxiv.org/abs/2312.10997)  

7. **Testing and Evaluation**  
- **Perspective A — A/B Testing:** Test multiple variants of the same prompt.  
  **Source:** [Google DeepMind Prompt Optimization Research](https://arxiv.org/abs/2309.03409)  
- **Perspective B — Guardrail Testing:** Check failure modes, hallucination hotspots, and misinterpretation risks.  
- **Perspective C — Automation:** Use scripts or evaluation frameworks (e.g., [Promptfoo](https://promptfoo.dev)) to benchmark prompts.

### Transformer

Transformers acts as the model-definition framework for state-of-the-art machine learning models in text, computer vision, audio, video, and multimodal model, for both inference and training.

It centralizes the model definition so that this definition is agreed upon across the ecosystem. transformers is the pivot across frameworks: if a model definition is supported, it will be compatible with the majority of training frameworks (Axolotl, Unsloth, DeepSpeed, FSDP, PyTorch-Lightning, …), inference engines (vLLM, SGLang, TGI, …), and adjacent modeling libraries (llama.cpp, mlx, …) which leverage the model definition from transformers.

We pledge to help support new state-of-the-art models and democratize their usage by having their model definition be simple, customizable, and efficient.

There are over 1M+ Transformers model checkpoints on the Hugging Face Hub you can use.

Explore the Hub today to find a model and use Transformers to help you get started right away.

Explore the Models Timeline to discover the latest text, vision, audio and multimodal model architectures in Transformers.
Features

Transformers provides everything you need for inference or training with state-of-the-art pretrained models. Some of the main features include:

- Pipeline: Simple and optimized inference class for many machine learning tasks like text generation, image segmentation, automatic speech recognition, document question answering, and more.
- Trainer: A comprehensive trainer that supports features such as mixed precision, torch.compile, and FlashAttention for training and distributed training for PyTorch models.
- generate: Fast text generation with large language models (LLMs) and vision language models (VLMs), including support for streaming and multiple decoding strategies.

https://huggingface.co/docs/transformers/index

## Agentic flow frameworks

- **Crewai**
    CrewAI is a lean, lightning-fast Python framework built entirely from scratch—completely independent of LangChain or other agent frameworks. CrewAI empowers developers with both high-level simplicity and precise low-level control, ideal for creating autonomous AI agents tailored to any scenario:

    - CrewAI Crews: Optimize for autonomy and collaborative intelligence, enabling you to create AI teams where each agent has specific roles, tools, and goals.
    - CrewAI Flows: Enable granular, event-driven control, single LLM calls for precise task orchestration and supports Crews natively.

    

- **Open sdk**
    The OpenAI Agents SDK enables you to build agentic AI apps in a lightweight, easy-to-use package with very few abstractions. It's a production-ready upgrade of our previous experimentation for agents, Swarm. The Agents SDK has a very small set of primitives:

    - Agents, which are LLMs equipped with instructions and tools
    - Handoffs, which allow agents to delegate to other agents for specific tasks
    - Guardrails, which enable validation of agent inputs and outputs
    - Sessions, which automatically maintains conversation history across agent runs

    In combination with Python, these primitives are powerful enough to express complex relationships between tools and agents, and allow you to build real-world applications without a steep learning curve. In addition, the SDK comes with built-in tracing that lets you visualize and debug your agentic flows, as well as evaluate them and even fine-tune models for your application.

## Project in UV

**Project structure**

A project consists of a few important parts that work together and allow uv to manage your project. In addition to the files created by uv init, uv will create a virtual environment and uv.lock file in the root of your project the first time you run a project command, i.e., uv run, uv sync, or uv lock.

A complete listing would look like:

```markdown
.
├── .venv
│   ├── bin
│   ├── lib
│   └── pyvenv.cfg
├── .python-version
├── README.md
├── main.py
├── pyproject.toml
└── uv.lock

pyproject.toml

The pyproject.toml contains metadata about your project:
pyproject.toml

[project]
name = "hello-world"
version = "0.1.0"
description = "Add your description here"
readme = "README.md"
dependencies = []

You'll use this file to specify dependencies, as well as details about the project such as its description or license. You can edit this file manually, or use commands like uv add and uv remove to manage your project from the terminal.

Tip

See the official pyproject.toml guide for more details on getting started with the pyproject.toml format.

You'll also use this file to specify uv configuration options in a [tool.uv] section.
.python-version

The .python-version file contains the project's default Python version. This file tells uv which Python version to use when creating the project's virtual environment.
.venv

The .venv folder contains your project's virtual environment, a Python environment that is isolated from the rest of your system. This is where uv will install your project's dependencies.

See the project environment documentation for more details.
uv.lock

uv.lock is a cross-platform lockfile that contains exact information about your project's dependencies. Unlike the pyproject.toml which is used to specify the broad requirements of your project, the lockfile contains the exact resolved versions that are installed in the project environment. This file should be checked into version control, allowing for consistent and reproducible installations across machines.

uv.lock is a human-readable TOML file but is managed by uv and should not be edited manually.

See the lockfile documentation for more details.

You could use uv pip install or uv add:

**Purpose & Persistence**
uv add <pkg>: This is the high-level dependency management command. It adds the package to your project’s pyproject.toml and updates the uv.lock file. Think of it as “declare this as a project dependency.” 
uv pip install <pkg>: This is more like “just install this now,” without modifying your project’s lock or pyproject.toml. Good for one-off installs or non‑persisted dependencies. 

**Environment & Locking**

When you run uv add, uv ensures your virtual environment (or project venv) is in sync with the lockfile, and then it installs. 
uv pip install bypasses that higher‑level dependency logic; it installs packages directly into the environment managed by uv, but doesn’t touch your lockfile. 

**Speed & Performance**
uv is very fast compared to pip because of its implementation in Rust and more efficient dependency resolution. 
by default, uv pip install does not compile bytecode (.pyc) at install time, which contributes to some of its speed. 

**Environment Creation**

uv manages virtual environments automatically. If there isn’t a .venv, uv can create one. 
Once you have that environment, uv add ensures your project dependencies are tracked (in pyproject.toml + uv.lock). 
For running code, you typically use uv run <command> so that you always run inside the correct environment.

**Xet: Storage Backend**

Repositories on the Hugging Face Hub are different from those on software development platforms. They contain files that are:

- Large - model or dataset files are in the range of GB and above. We have a few TB-scale files!
- Binary - not in a human readable format by default (e.g., [Safetensors](https://huggingface.co/docs/safetensors/en/index) or [Parquet](https://huggingface.co/docs/dataset-viewer/en/parquet#what-is-parquet))

While the Hub leverages modern version control with the support of Git, these differences make [Model](https://huggingface.co/docs/hub/models) and [Dataset](https://huggingface.co/docs/hub/datasets) repositories quite different from those that contain only source code.

Storing these files directly in a pure Git repository is impractical. Not only are the typical storage systems behind Git repositories unsuited for such files, but when you clone a repository, Git retrieves the entire history, including all file revisions. This can be prohibitively large for massive binaries, forcing you to download gigabytes of historic data you may never need.

Instead, on the Hub, these large files are tracked using "pointer files" and identified through a `.gitattributes` file (both discussed in more detail below), which remain in the Git repository while the actual data is stored in remote storage (like [Amazon S3](https://aws.amazon.com/s3/)). As a result, the repository stays small and typical Git workflows remain efficient.




Historically, Hub repositories have relied on [Git LFS](https://git-lfs.com/) for this mechanism. While Git LFS remains supported (see [Backwards Compatibility & Legacy](./legacy-git-lfs)), the Hub has adopted Xet, a modern custom storage system built specifically for AI/ML development. It enables chunk-level deduplication, smaller uploads, and faster downloads than Git LFS.

**Recommendations**

Xet integrates seamlessly with all of the Hub's workflows. However, there are a few steps you may consider to get the most benefits from Xet storage.

When uploading or downloading with Python:

- **Make sure `hf_xet` is installed**: While Xet remains backward compatible with legacy clients optimized for Git LFS, the `hf_xet` integration with `huggingface_hub` delivers optimal chunk-based performance and faster iteration on large files.
- **Utilize `hf_xet` environment variables**: The default installation of `hf_xet` is designed to support the broadest range of hardware. To take advantage of setups with more network bandwidth or processing power read up on `hf_xet`'s [environment variables](https://huggingface.co/docs/huggingface_hub/package_reference/environment_variables#xet) to optimize downloads and uploads.

When uploading or downloading in Git or Python: 

- **Leverage frequent, incremental commits**: Xet's chunk-level deduplication means you can safely make incremental updates to models or datasets. Only changed chunks are uploaded, so frequent commits are both fast and storage-efficient.
- **Be Specific in .gitattributes**: When defining patterns for Xet or LFS, use precise file extensions (e.g., `*.safetensors`, `*.bin`) to avoid unnecessarily routing smaller files through large-file storage.
- **Prioritize community access**: Xet substantially increases the efficiency and scale of large file transfers. Instead of structuring your repository to reduce its total size (or the size of individual files), organize it for collaborators and community users so they may easily navigate and retrieve the content they need.

**Current Limitations**

While Xet brings fine-grained deduplication and enhanced performance to Git-based storage, some features and platform compatibilities are still in development. As a result, keep the following constraints in mind when working with a Xet-enabled repository:

- **64-bit systems only**: Both `hf_xet` and Git Xet currently require a 64-bit architecture; 32-bit systems are not supported.

````python
"""
Enhanced LM Studio Qwen-VL Example
----------------------------------
This script uses LM Studio's AsyncClient to:
1. Load a Qwen-VL model (e.g., "qwen/qwen3-vl-30b").
2. Send an image for multimodal reasoning (OCR + translation).
3. Control generation behavior (temperature, top_p, max_tokens).
4. Print the model's prediction.

Temperature = 0.1  (low randomness, more deterministic)
"""

import lmstudio as lms
import asyncio


async def process_image_with_model(image_path: str, model_name: str, temperature: float = 0.1):
    """
    Processes an image through the Qwen-VL model with specified temperature.

    Args:
        image_path (str): Path to the input image file.
        model_name (str): LM Studio model identifier (e.g., "qwen/qwen3-vl-30b").
        temperature (float): Sampling temperature (0.0 = deterministic, >1 = more random).

    Returns:
        str: Model-generated response text.
    """
    # Initialize LM Studio asynchronous client
    async with lms.AsyncClient() as client:
        # Step 1: Prepare the image for upload
        image_handle = await client.files.prepare_image(image_path)

        # Step 2: Load the model
        model = await client.llm.model(model_name)

        # Step 3: Create a new chat session
        chat = lms.Chat()
        chat.add_user_message(
            (
                """Can you identify the language? The expected output is ONE WORD
                (the language name in <angle brackets>, e.g. <English>).
                After that, extract the exact text in the original language 
                including sound effects. Then translate all text to English concisely
                and precisely, without adding or guessing words."""
            ),
            images=[image_handle],
        )

        # Step 4: Set response generation configuration
        # config = {
        #     "temperature": temperature,  # controls randomness of output
        #     "top_p": 0.9,                # nucleus sampling parameter
        #     "max_tokens": 512,           # safety cap on response length
        # }
        config = {
            "temperature": temperature,  # controls randomness of output
            
        }

        # Step 5: Send the request and await the prediction
        prediction = await model.respond(chat, config=config)

        # Step 6: Return the response text
        return prediction


async def main():
    """
    Main coroutine to execute the LM Studio Qwen-VL image OCR + translation pipeline.
    """
    
    image_path = "/Users/alexgreco/projects/manga_translate/raw/8-fb166.jpg"
    model_name = "qwen/qwen3-vl-30b"

    try:
        result = await process_image_with_model(image_path, model_name, temperature=0.1)
        print("\n--- Model Response ---\n")
        print(result)
        print("\n----------------------\n")
    except Exception as e:
        print(f"❌ Error: {e}")



# if __name__ == "__main__":
#     asyncio.run(main())
````

````python
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
import os

path = "/Users/alexgreco/projects/ai_basic/images"

def generate_image(prompt, filename="{path}/generated_image.png", num_inference_steps=20):
    """
    Generate an image from a text prompt using Stable Diffusion
    
    Args:
        prompt (str): Text description of the image you want to generate
        filename (str): Output filename
        num_inference_steps (int): Number of denoising steps (more steps = better quality but slower)
    """
    
    # Check if GPU is available
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Using device: {device}")
    
    # Load the Stable Diffusion pipeline
    # Using a smaller model for faster generation and less memory usage
    model_id = "runwayml/stable-diffusion-v1-5"
    
    pipe = StableDiffusionPipeline.from_pretrained(
        model_id,
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
        use_safetensors=True
    )
    pipe = pipe.to(device)
    
    # Generate the image
    print(f"Generating image for prompt: '{prompt}'")
    with torch.no_grad():
        image = pipe(
            prompt=prompt,
            num_inference_steps=num_inference_steps,
            guidance_scale=7.5  # Controls how closely to follow the prompt
        ).images[0]
    
    # Save the image
    image.save(filename)
    print(f"Image saved as {filename}")
    return image

# Example usage
if __name__ == "__main__":
    prompt = "A beautiful sunset over mountains, digital art"
    generated_image = generate_image(prompt, "{path}/sunset_mountains.png")
    generated_image.show()  # This will open the image in your default viewer
````

````python
import torch
from diffusers import DiffusionPipeline
from PIL import Image

def generate_image_fast(prompt, filename="fast_generated_image.png"):
    """
    Generate images using a smaller, faster model
    """
    device = "cuda" if torch.cuda.is_available() else "cpu"
    
    # Use a smaller model
    pipe = DiffusionPipeline.from_pretrained(
        "OFA-Sys/small-stable-diffusion-v0",
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    )
    pipe = pipe.to(device)
    
    # Generate image
    image = pipe(prompt).images[0]
    image.save(filename)
    print(f"Image saved as {filename}")
    return image

# Example usage
if __name__ == "__main__":
    prompt = "A cute cartoon cat playing with a ball"
    generate_image_fast(prompt, "cartoon_cat.png")
````

````python
import torch
from diffusers import StableDiffusionPipeline, DPMSolverMultistepScheduler
from PIL import Image

def generate_image_advanced(
    prompt, 
    negative_prompt="",
    filename="advanced_image.png",
    width=512,
    height=512,
    num_inference_steps=20,
    guidance_scale=7.5,
    seed=None
):
    """
    Generate images with advanced configuration options
    """
    device = "cuda" if torch.cuda.is_available() else "cpu"
    
    pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
        use_safetensors=True
    )
    
    # Use a faster scheduler
    pipe.scheduler = DPMSolverMultistepScheduler.from_config(pipe.scheduler.config)
    pipe = pipe.to(device)
    
    # Set seed for reproducibility
    if seed is not None:
        torch.manual_seed(seed)
    
    # Generate image
    with torch.no_grad():
        image = pipe(
            prompt=prompt,
            negative_prompt=negative_prompt,
            width=width,
            height=height,
            num_inference_steps=num_inference_steps,
            guidance_scale=guidance_scale,
        ).images[0]
    
    image.save(filename)
    print(f"Image saved as {filename}")
    return image

# Example usage
if __name__ == "__main__":
    prompt = "A majestic dragon flying over a medieval castle, fantasy art"
    negative_prompt = "blurry, low quality, distorted"
    
    generate_image_advanced(
        prompt=prompt,
        negative_prompt=negative_prompt,
        filename="dragon_castle.png",
        width=768,
        height=512,
        seed=42  # For reproducible results
    )
````

````python
def generate_multiple_images(prompts, filenames=None):
    """
    Generate multiple images from a list of prompts
    """
    device = "cuda" if torch.cuda.is_available() else "cpu"
    
    pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    )
    pipe = pipe.to(device)
    
    if filenames is None:
        filenames = [f"image_{i}.png" for i in range(len(prompts))]
    
    images = []
    for i, prompt in enumerate(prompts):
        print(f"Generating image {i+1}/{len(prompts)}: '{prompt}'")
        with torch.no_grad():
            image = pipe(prompt).images[0]
            image.save(filenames[i])
            images.append(image)
    
    print("All images generated successfully!")
    return images

# Example usage
if __name__ == "__main__":
    prompts = [
        "A serene lake at sunrise",
        "A futuristic city at night",
        "An astronaut riding a horse on Mars"
    ]
    
    filenames = ["lake.png", "city.png", "astronaut.png"]
    generate_multiple_images(prompts, filenames)
````

````python
# Searchable Transcript Archive


import sqlite3

audio_path = "/Users/alexgreco/projects/ai_basic/audio_to_text.m4a"

# 1. Transcribe
result = model.transcribe(audio_path)

# 2. Connect to DB
conn = sqlite3.connect("/Users/alexgreco/projects/ai_basic/audio_index.db")
conn.execute("""
CREATE TABLE IF NOT EXISTS transcripts (
    id INTEGER PRIMARY KEY, 
    text TEXT, 
    start REAL, 
    end REAL
)
""")

# 3. Insert transcript (fallback if segments not present)
if "segments" in result:
    for seg in result["segments"]:
        conn.execute(
            "INSERT INTO transcripts (text, start, end) VALUES (?, ?, ?)",
            (seg["text"], seg["start"], seg["end"])
        )
else:
    # Insert full text as single entry
    conn.execute(
        "INSERT INTO transcripts (text, start, end) VALUES (?, ?, ?)",
        (result["text"], 0.0, 0.0)
    )

conn.commit()

# 4. Search example
query = "BCP"
rows = conn.execute(
    "SELECT start, end, text FROM transcripts WHERE text LIKE ?",
    (f"%{query}%",)
).fetchall()

if rows:
    for start, end, text in rows:
        print(f"[{start:.2f}s → {end:.2f}s] {text}")
else:
    print(f"No results found for query: '{query}'")

conn.close()

````

````python
def to_srt(segments, path="/Users/alexgreco/projects/ai_basic/subtitles.srt"):
    with open(path, "w") as f:
        for i, seg in enumerate(segments, 1):
            start = f"{int(seg['start'] // 3600):02}:{int(seg['start'] % 3600 // 60):02}:{seg['start'] % 60:06.3f}".replace('.', ',')
            end = f"{int(seg['end'] // 3600):02}:{int(seg['end'] % 3600 // 60):02}:{seg['end'] % 60:06.3f}".replace('.', ',')
            f.write(f"{i}\n{start} --> {end}\n{seg['text'].strip()}\n\n")

result = model.transcribe(audio_path)
to_srt(result["segments"])
print("✅ SRT subtitles created.")
````

````python
#function to add automatically from database csv

def add_values(collection,file_name,document_column, id_colum):
  # Reading a CSV file with a different delimiter (e.g., semicolon)
  df = pd.read_csv(file_name, delimiter='|')
  df=df[0:100]

  final=[]


  for _, row in df.iterrows():
      res_dict = {}
      res_all = ""
      target_text=row[document_column]
      if tokens(target_text)<999:
        target_text=""
      id_col=row[id_colum]


      for _,column in enumerate(df.columns):
        #dic with all the metadata
        if column!=id_colum:
          res_all = res_all + str(column) + " = " + str(row[column]) + " "
          if tokens(res_all)<999:
            res_all=""
          if column!=document_column:
            res_dict[column]=row[column]


      final.append(res_dict)



      if collection == 'collection_chroma':
        collection_chroma.add(
                      documents = str(target_text),
                      metadatas = final,
                      ids = str(id_col)
                      )
      elif collection == 'collection_palm':
        collection_palm.add(
                      documents = str(target_text),
                      metadatas = final,
                      ids = str(id_col)
                      )
      elif collection == 'collection_chroma_all':
        collection_chroma_all.add(
                      documents = str(res_all),
                      metadatas = final,
                      ids = str(id_col)
                      )
      elif collection == 'collection_palm_all':
        collection_palm_all.add(
                      documents = str(res_all),
                      metadatas = final,
                      ids = str(id_col)
                      )


      final=[]

  return final
````

````python
def make_prompt(query, relevant_passage):
  final=''
  for x in relevant_passage:
    escaped = x.replace("'", "").replace('"', "").replace("\n", " ")
    final+=escaped

  prompt = ("""You are a helpful and informative bot that answers questions using text from the reference passage included below. \
  Be sure to respond in a complete sentence, being comprehensive, including all relevant background information. \
  However, you are talking to a non-technical audience, so be sure to break down complicated concepts and \
  strike a friendly and converstional tone. \
  If the passage is irrelevant to the answer, you may ignore it.
  QUESTION: '{query}'
  PASSAGE: '{relevant_passage}'

    ANSWER:
  """).format(query=query, relevant_passage=final)

  print(f"this is the promt {prompt}")
  return prompt
````

The `answer` function will generate a response based on the query you have passed in. It retrieves the relevant document, and from there calls the PaLM text generation API to generate a response to the query.

````python
text_models = [m for m in palm.list_models() if 'generateText' in m.supported_generation_methods]
text_model = text_models[0]
text_model
````

````python
def answer(model, query, db, temperature=0.5):
  passage = get_relevant_passage(query, db)
  prompt = make_prompt(query, passage)
  answer = palm.generate_text(prompt=prompt, model=model, candidate_count=3, temperature=temperature, max_output_tokens=1000)
  return answer
  # return answer.candidates[0]['output']
````

The Model Context Protocol (MCP) has revolutionized how AI models interact with external data sources, but as adoption grows, so do the challenges around token efficiency. In this article, I’ll share how we achieved a dramatic 70–80% reduction in token usage by reimagining our MCP server architecture with GraphQL, transforming a token-hungry system into an efficient, scalable solution.
The Problem: Token Bloat in Traditional MCP Implementations

When we first implemented MCP servers for AI data access, our approach seemed straightforward: wrap our existing API layer and expose each endpoint as an individual MCP tool. This design offered several advantages:

    Authorization handled at the API level: Existing entitlement and auth mechanisms worked seamlessly
    Familiar patterns: Developers could leverage existing API knowledge
    Rapid deployment: Minimal refactoring of existing services

However, this approach quickly revealed a critical flaw: token inefficiency.
The Token Usage Crisis

Our initial implementation suffered from two primary issues:

    Verbose API responses: Each tool call returned complete JSON objects, often containing unnecessary nested data and metadata that consumed significant tokens
    Tool proliferation: Breaking large responses into smaller chunks meant creating numerous individual MCP tools, leading to:

    Complex orchestration across multiple tool calls
    Increased context switching overhead
    Higher total token consumption due to repeated metadata

Consider this example scenario:

// Single API call returning user data
{
  "user": {
    "id": "12345",
    "profile": {
      "name": "John Doe",
      "email": "john@example.com",
      "preferences": { /* 50+ fields */ },
      "metadata": { /* extensive tracking data */ }
    },
    "permissions": [ /* 20+ permission objects */ ],
    "usage_stats": { /* detailed analytics */ }
  }
}

When the AI model only needed the user’s name and email, we were still consuming tokens for the entire response payload — a classic case of over-fetching leading to token waste.
The GraphQL Solution: Two Tools to Rule Them All

Our breakthrough came from recognizing that GraphQL’s core strength — precise data fetching — directly addresses MCP’s token efficiency challenge. Instead of exposing dozens of individual API endpoints as MCP tools, we consolidated our entire data layer into just two powerful tools:
Tool 1: Schema Introspection (get_graphql_schema)

def get_graphql_schema():
    """Returns the complete GraphQL schema definition"""
    return {
        "schema": schema_definition_string,
        "types": available_types,
        "queries": available_queries,
        "mutations": available_mutations
    }

Tool 2: Query Execution (execute_graphql_query)

def execute_graphql_query(query: str, variables: dict = None):
    """Executes a GraphQL query and returns precisely requested data"""
    result = graphql_executor.execute(query, variables)
    return result.data

Schema Design Strategy

The key insight was treating our GraphQL schema as a representation layer for existing APIs rather than a complete reimplementation. Here’s how we mapped our REST endpoints to GraphQL:

-- REST: GET /users/{id}
Becomes:

type User {
  id: ID!
  name: String!
  email: String!
  profile: UserProfile
  permissions: [Permission!]!
}

-- REST: GET /orders/{userId}
Becomes:

type Query {
  user(id: ID!): User
  userOrders(userId: ID!): [Order!]!
}

This approach preserved our existing authorization logic while enabling precise data fetching.

def generate_graphql_schema_from_openapi(openapi_spec):
    """Convert OpenAPI definitions to GraphQL schema"""
    schema_builder = GraphQLSchemaBuilder()
    
    for path, methods in openapi_spec['paths'].items():
        for method, definition in methods.items():
            if method == 'get':
                # Convert to GraphQL Query
                schema_builder.add_query(path, definition)
            elif method == 'post':
                # Convert to GraphQL Mutation
                schema_builder.add_mutation(path, definition)
    
    return schema_builder.build()

The PARAMETER instruction defines a parameter that can be set when the model is run.

Copy
PARAMETER <parameter> <parametervalue>

Valid Parameters and Values
Parameter	Description	Value Type	Example Usage
mirostat	Enable Mirostat sampling for controlling perplexity. (default: 0, 0 = disabled, 1 = Mirostat, 2 = Mirostat 2.0)	int	mirostat 0
mirostat_eta	Influences how quickly the algorithm responds to feedback from the generated text. A lower learning rate will result in slower adjustments, while a higher learning rate will make the algorithm more responsive. (Default: 0.1)	float	mirostat_eta 0.1
mirostat_tau	Controls the balance between coherence and diversity of the output. A lower value will result in more focused and coherent text. (Default: 5.0)	float	mirostat_tau 5.0
num_ctx	Sets the size of the context window used to generate the next token. (Default: 2048)	int	num_ctx 4096
repeat_last_n	Sets how far back for the model to look back to prevent repetition. (Default: 64, 0 = disabled, -1 = num_ctx)	int	repeat_last_n 64
repeat_penalty	Sets how strongly to penalize repetitions. A higher value (e.g., 1.5) will penalize repetitions more strongly, while a lower value (e.g., 0.9) will be more lenient. (Default: 1.1)	float	repeat_penalty 1.1
temperature	The temperature of the model. Increasing the temperature will make the model answer more creatively. (Default: 0.8)	float	temperature 0.7
seed	Sets the random number seed to use for generation. Setting this to a specific number will make the model generate the same text for the same prompt. (Default: 0)	int	seed 42
stop	Sets the stop sequences to use. When this pattern is encountered the LLM will stop generating text and return. Multiple stop patterns may be set by specifying multiple separate stop parameters in a modelfile.	string	stop “AI assistant:“
num_predict	Maximum number of tokens to predict when generating text. (Default: -1, infinite generation)	int	num_predict 42
top_k	Reduces the probability of generating nonsense. A higher value (e.g. 100) will give more diverse answers, while a lower value (e.g. 10) will be more conservative. (Default: 40)	int	top_k 40
top_p	Works together with top-k. A higher value (e.g., 0.95) will lead to more diverse text, while a lower value (e.g., 0.5) will generate more focused and conservative text. (Default: 0.9)	float	top_p 0.9
min_p	Alternative to the topp, and aims to ensure a balance of quality and variety. The parameter _p represents the minimum probability for a token to be considered, relative to the probability of the most likely token. For example, with p=0.05 and the most likely token having a probability of 0.9, logits with a value less than 0.045 are filtered out. (Default: 0.0)	float	min_p 0.05

### errors

Endpoints return appropriate HTTP status codes based on the success or failure of the request in the HTTP status line (e.g. HTTP/1.1 200 OK or HTTP/1.1 400 Bad Request). Common status codes are:

    200: Success
    400: Bad Request (missing parameters, invalid JSON, etc.)
    404: Not Found (model doesn't exist, etc.)
    429: Too Many Requests (e.g. when a rate limit is exceeded)
    500: Internal Server Error
    502: Bad Gateway (e.g. when a cloud model cannot be reached)


Error messages

Errors are returned in the application/json format with the following structure, with the error message in the error property:

{
  "error": "the model failed to generate a response"
}

Errors that occur while streaming

If an error occurs mid-stream, the error will be returned as an object in the application/x-ndjson format with an error property. Since the response has already started, the status code of the response will not be changed.

{"model":"gemma3","created_at":"2025-10-26T17:21:21.196249Z","response":" Yes","done":false}
{"model":"gemma3","created_at":"2025-10-26T17:21:21.207235Z","response":".","done":false}
{"model":"gemma3","created_at":"2025-10-26T17:21:21.219166Z","response":"I","done":false}
{"model":"gemma3","created_at":"2025-10-26T17:21:21.231094Z","response":"can","done":false}
{"error":"an error was encountered while running the model"}

### streaming

Certain API endpoints stream responses by default, such as /api/generate. These responses are provided in the newline-delimited JSON format (i.e. the application/x-ndjson content type). For example:

Disabling streaming

Streaming can be disabled by providing {"stream": false} in the request body for any endpoint that support streaming. This will cause responses to be returned in the application/json format instead:

{"model":"gemma3","created_at":"2025-10-26T17:15:24.166576Z","response":"That's a fantastic question!","done":true}

When to use streaming vs non-streaming

Streaming (default):

    Real-time response generation
    Lower perceived latency
    Better for long generations

Non-streaming:

    Simpler to process
    Better for short responses, or structured outputs
    Easier to handle in some applications

````python
from airllm import AutoModel

MAX_LENGTH = 128
# could use hugging face model repo id:
model = AutoModel.from_pretrained("unsloth/Meta-Llama-3.1-405B-Instruct-bnb-4bit")

input_text = [
        'What is the capital of United States?',
    ]

input_tokens = model.tokenizer(input_text,
    return_tensors="pt", 
    return_attention_mask=False, 
    truncation=True, 
    max_length=MAX_LENGTH, 
    padding=False)
           
generation_output = model.generate(
    input_tokens['input_ids'].cuda(), 
    max_new_tokens=10,
    use_cache=True,
    return_dict_in_generate=True)

output = model.tokenizer.decode(generation_output.sequences[0])

print(output)
````
